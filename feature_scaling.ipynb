```python
import numpy as np
```

```python
np.random.seed(861)      #for reproducibility
X = np.random.randint(1, 100000, size=5)
print(X)
```

## Standardization using scikit-learn

```python
from sklearn.preprocessing import StandardScaler
s = StandardScaler()
X_scaled = s.fit_transform(X.reshape(-1,1))
print(X_scaled)
```

## Standardization from Scratch

```python
# mean = np.mean(data)
# print(mean)

s = 0
for i in X:
    s+= i

mean = s/len(X)
print(mean)
```

```python
# sdeviation = np.std(data)
# print(sdeviation)
sq = 0
for i in X:
    sq = sq + ((i-mean)**2)

std = (sq/len(X))**(1/2)
print(std)
```

```python
X_scaled = []
for i in X:
    X_scaled.append((i-mean)/std)
    
print(X_scaled)
```

## Normalization using scikit-learn

```python
from sklearn.preprocessing import MinMaxScaler
m = MinMaxScaler()
X_scaled2 = m.fit_transform(X.reshape(-1,1))
print(X_scaled2)
```
