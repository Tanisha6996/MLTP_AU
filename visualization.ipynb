## Line Charts

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame({'Values': [1,2,4,6,5,4,3,2,1,10], 
                   'Months': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct']
                              })
```

```python
plt.plot(df['Months'], df['Values'])
```

```python
# Sample data
days = [1, 2, 3, 4, 5, 6, 7]
temperature = [28, 30, 31, 33, 32, 31, 29]


plt.plot(
    days,                     # x-axis values
    temperature,              # y-axis values
    color='yellow',              # line color
    linestyle='--',           # line style: '-', '--', '-.', ':'
    linewidth=2,              # thickness of line
    marker='o',               # marker at data points
    markersize=8,             # size of marker
    markerfacecolor='green',   # fill color of marker
    markeredgecolor='black',  # border color of marker
    alpha=0.8,                # transparency (0–1)
    label='Daily Temperature' # legend label
)

# Titles and labels
plt.title("Temperature Variation Over Days", fontsize=14)
plt.xlabel("Day", fontsize=12)
plt.ylabel("Temperature (°C)", fontsize=12)

# Axis customization
plt.xticks(days)
plt.ylim(25, 35)

# Grid
#plt.grid(True)

# Legend
#plt.legend()

# Show plot
plt.show()
```

## Bar Charts

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample categorical data
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [250, 400, 150, 300]
```

```python
plt.bar(products, sales)
plt.show()
```

```python
plt.bar(
    products,              # categories (x-axis)
    sales,                 # values (y-axis)
    color='lightpink',       # bar color
    edgecolor='black',     # border color
    width=0.6,             # width of bars
    alpha=0.8,             # transparency
    label='Product Sales'  # legend label
)

# Titles and axis labels
plt.title("Sales by Product", fontsize=14)
plt.xlabel("Products", fontsize=12)
plt.ylabel("Sales (Units)", fontsize=12)

# Axis ticks
plt.xticks(rotation=90)

# Grid (usually on y-axis)
plt.grid(axis='y', linestyle='--', alpha=0.6)

# Legend
plt.legend()

# Display the plot
plt.show()
```

## Histograms

```python
# Generate random numerical data (normal distribution)
np.random.seed(42)          # for reproducibility- Getting the same results every time you run the code.
marks = np.random.normal(loc=70, scale=10, size=200) 
#It generates 200 random numbers that follow a normal (bell-shaped) distribution with an average of 70 and a spread of 10.
```

```python
plt.hist(marks)
plt.show()
```

```python
plt.hist(
    marks,                 # generated numerical data
    bins=12,               # number of bins
    color='lightgreen',    # bar color
    edgecolor='black',     # border color
    alpha=0.75,            # transparency
    histtype='bar',        # bar / step / stepfilled
    orientation='horizontal',
    rwidth=0.9,            # relative width of bars
    density=False,         # frequency count
    label='Marks Distribution'
)

# Titles and labels
plt.title("Histogram of Randomly Generated Marks", fontsize=14)
plt.xlabel("Marks", fontsize=12)
plt.ylabel("Frequency", fontsize=12)

# Grid
plt.grid(axis='y', linestyle='--', alpha=0.6)

# Legend
plt.legend()

# Show plot
plt.show()
```

## Scatter Plot

```python
# Generate sample numerical data
np.random.seed(42)
age = np.random.randint(18, 60, 50)
salary = np.random.randint(20000, 100000, 50)
```

```python
plt.scatter(age, salary)
plt.show()
```

```python
plt.scatter(
    age,                     # x-axis values
    salary,                  # y-axis values
    c=salary,                # color mapped to values
    cmap='viridis',          # color map
    s=10,                   # size of points
    alpha=0.7,               # transparency
    marker='o',              # marker shape
    edgecolors='black',      # marker border color
    linewidths=0.8,          # border thickness
    label='Employee Data'    # legend label
)

# Titles and axis labels
plt.title("Age vs Salary Scatter Plot", fontsize=14)
plt.xlabel("Age", fontsize=12)
plt.ylabel("Salary", fontsize=12)

# Grid
plt.grid(True, linestyle='--', alpha=0.6)

# Colorbar (important when using cmap)
plt.colorbar(label="Salary Level")

# Legend
plt.legend()

# Show plot
plt.show()
```

## Area Plot

```python
# Sample data
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [200, 300, 450, 600, 750, 900]

plt.figure(figsize=(8, 5))

plt.fill_between(
    months,              # x-axis
    sales,               # y-axis
    color='skyblue',     # fill color
    alpha=0.6,           # transparency
    label='Sales Growth'
)

plt.plot(
    months,
    sales,
    color='blue',
    linewidth=2
)

plt.title("Cumulative Sales Over Months")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.legend()
plt.show()
```

## Pie charts

```python
labels = ['Food', 'Rent', 'Transport', 'Savings']
expenses = [40, 30, 15, 15]

plt.figure(figsize=(6, 6))

plt.pie(
    expenses,
    labels=labels,
    autopct='%1.1f%%',   # percentage format
    startangle=90,       # rotate pie
    explode=[0.1, 0, 0, 0],  # highlight slice
    shadow=True
)

plt.title("Monthly Expense Distribution")
plt.show()
```

## HeatMap

```python
# Sample data
data = np.random.rand(6, 6)

plt.figure(figsize=(6, 5))

plt.imshow(
    data,
    cmap='coolwarm',     # color map
    interpolation='nearest'
)

plt.colorbar(label='Value Intensity')

plt.title("Heatmap Example")
plt.xticks(range(6))
plt.yticks(range(6))
plt.show()
```
