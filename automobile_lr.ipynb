## Now let's understand this pattern,

\s*\d+\.\s*([a-zA-Z0-9\-]+):

1. \s* - Handles indentation before numbers (Matches zero or more spaces)
2. \d - Matches one of more digits
3. \. - Matches a literal dot
4. \s* - Handles indentation after dot and before the column name (text)
5. ([a-zA-Z0-9\-]+) - It captures: Letters (a-z, A-Z),  Numbers (0-9), Hyphen (-)
6. This has three important components: [] → character set,  + → quantifier (Match one or more of the previous thing.),  () → capturing group
7. match.group(1) extracts what was inside parentheses.

```python
# Fill numerical missing values with median
for col in df.select_dtypes(include="number").columns:
    df[col] = df[col].fillna(df[col].median())

# Fill categorical with mode if any
for col in df.select_dtypes(exclude="number").columns:
    df[col] = df[col].fillna(df[col].mode()[0])

print("Missing after fill:", df.isnull().sum().sum())
```

```python
df["Date"] = pd.to_datetime(df["Formatted Date"])
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day

df.drop(["Formatted Date", "Date"], axis=1, inplace=True)
```
